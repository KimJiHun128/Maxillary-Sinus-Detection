```python
#  데이터셋 준비 
from sklearn.datasets import fetch_openml 
mnist = fetch_openml('mnist_784', version=1, as_frame=False) 
X, y = mnist["data"], mnist["target"]
X = X / 255.0  # 픽셀 스케일링
#학습을 위하여 문자로 표시된 label값을 숫자로 변경
y = y.astype(np.uint8) # string을 unsigned int로 변환
'''
# 처음 5만개는 학습에 그다음 만개는 val,마지막 만개는 테스트에 사용 
X_train,X_val, X_test, = X[:50000], X[50000:60000],X[60000:]
y_train,y_val, y_test =  y[:50000], y[50000:60000],y[60000:]
'''
# 처음 6만개는 학습에 다음 만개는 테스트로 사용 
X_train, X_test, y_train, y_test = X[:60000], X[60000:], y[:60000], y[60000:]
```

```python
# PCA를 사용하여 특징을 154개로 압축 
from sklearn.decomposition import PCA 
pca = PCA(n_components = 154) 
X_reduced = pca.fit_transform(X_train) 
X_recovered = pca.inverse_transform(X_reduced)
```

```python
# RBF 커널 추가  gamma = 0.001
#감마는 함수폭을 조절 c는 허용정도를 조절 
from sklearn.svm import SVC
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

gamma1, gamma2,gamma3 = 0.001, 0.01, 1
C1, C2, C3  =1, 10, 100
# hyperparams = (gamma1, C1), (gamma1, C2),(gamma1, C3), (gamma2, C1), (gamma2, C2),(gamma2, C3), (gamma3, C1), (gamma3, C2),(gamma3, C3)
hyperparams =  (gamma1, C1), (gamma1, C2),(gamma1, C3)
for gamma, C in hyperparams :
    rbf_kernel_svm_clf = Pipeline([
            ("scaler", StandardScaler()),
            ("svm_clf", SVC(kernel="rbf", gamma=gamma, C=C))
        ])
    rbf_kernel_svm_clf.fit(X_recovered, y_train)
    

    # 정확도 평가
    from sklearn.metrics import accuracy_score
    # 모델을 사용하여 테스트 데이터로 예측 수행
    y_pred = rbf_kernel_svm_clf.predict(X_test)

    accuracy = accuracy_score(y_test, y_pred)
    print("gamma:",gamma,"C=",C)
    print("테스트 데이터의 정확도:", accuracy)
```
